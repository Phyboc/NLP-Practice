Given two sentences where one is positive sentence and other is negative sentence, calculate the probability sore using N-grams (1-grams, 2-grams, ...) and write the code for the using NLTK. (Keyword: Dress)

1. I like wearing simple dress.
2. I don't like green coloured dress.   

```python
from nltk.tokenize import word_tokenize
from nltk import ngrams
from collections import defaultdict

corpus = [
    "I like wearing simple dress",
    "I donot like green coloured dress"
]

tokenized_corpus = []
for sentence in corpus:
    tokens = word_tokenize(sentence)
    tokenized_corpus.append(tokens)

unigrams_list = []
for tokens in tokenized_corpus:
    unigrams = list(ngrams(tokens, 1))
    unigrams_list.append(unigrams)
bigrams_list = []
for tokens in tokenized_corpus: 
    bigrams = list(ngrams(tokens, 2))
    bigrams_list.append(bigrams)
trigrams_list = []
for tokens in tokenized_corpus:
    trigrams = list(ngrams(tokens, 3))
    trigrams_list.append(trigrams)


vocabulary = set()
for tokens in tokenized_corpus:
    vocabulary.update(tokens)
vocab_len = len(vocabulary)

word_count = {}

total_words = 0

for sentence in tokenized_corpus:

    for word in sentence:

        total_words += 1

        if word not in word_count:
            word_count[word] = 0

        word_count[word] += 1

print("\nWord Counts")

print(word_count)

print("Total Words =",total_words)

probability = {}

for word in vocabulary:

    count = word_count.get(word,0)

    prob = (count+1)/(total_words+vocab_len)

    probability[word] = prob

print("\nProbabilities")

for word in probability:

    print(word,"=",round(probability[word],3))


test = "I like wearing coloured dress"

test_words = test.split()

sentence_probability = 1

for word in test_words:

    sentence_probability *= probability[word]

print("\nSentence Probability =",sentence_probability)


N = len(test_words)

perplexity = (1/sentence_probability)**(1/N)

print("Perplexity =",round(perplexity,3))

```

```python
# Corpus
corpus = [
    "I like wearing simple dress",
    "I dont like green coloured dress"
]

sentences = []

for sentence in corpus:
    words = sentence.split()
    words.insert(0, "<s>")
    words.append("</s>")
    sentences.append(words)

print("Sentences")
print(sentences)

vocabulary = set()

for sentence in sentences:
    for word in sentence:
        vocabulary.add(word)

vocabulary = sorted(vocabulary)

print("\nVocabulary")
print(vocabulary)

V = len(vocabulary)

print("Vocabulary Size =", V)

bigram_count = {}

for sentence in sentences:

    for i in range(len(sentence)-1):

        first = sentence[i]
        second = sentence[i+1]

        if first not in bigram_count:
            bigram_count[first] = {}

        if second not in bigram_count[first]:
            bigram_count[first][second] = 0

        bigram_count[first][second] += 1

print("\nBigram Counts")

for first in bigram_count:
    print(first, ":", bigram_count[first])

row_total = {}

for first in bigram_count:

    total = 0

    for second in bigram_count[first]:
        total += bigram_count[first][second]

    row_total[first] = total

print("\nRow Totals")

print(row_total)

probability = {}

for first in vocabulary:

    probability[first] = {}

    history_count = row_total.get(first,0)

    for second in vocabulary:

        count = 0

        if first in bigram_count:

            count = bigram_count[first].get(second,0)

        prob = (count+1)/(history_count+V)

        probability[first][second] = prob

print("\nProbability Table")

for first in probability:

    print(first)

    for second in probability[first]:
        print(second,"=",round(probability[first][second],3))

test = "I like wearing coloured dress"

test_words = test.split()

test_words.insert(0,"<s>")
test_words.append("</s>")

print("\nTest Sentence")

print(test_words)

sentence_probability = 1

for i in range(len(test_words)-1):

    first = test_words[i]
    second = test_words[i+1]

    prob = probability[first][second]

    print(first,"->",second,"=",round(prob,3))

    sentence_probability *= prob

print("\nSentence Probability =",sentence_probability)


N = len(test_words)-1

perplexity = (1/sentence_probability)**(1/N)

print("Perplexity =",round(perplexity,3))
```

In this activity, we learnt how the probability of a word occurring in a sentence can be calculated using both Unigram and Bigram language models. In the Unigram model, the probability of a word depends only on its frequency of occurrence, whereas in the Bigram model, the probability of a word depends on the previous word in the sentence. We also learnt how Laplace (Add-1) smoothing is used to assign non-zero probabilities to unseen word combinations.

We further learnt that the perplexity score measures how well a language model predicts the next word in a sentence. A lower perplexity score indicates that the model predicts the sentence more accurately with less uncertainty, while a higher perplexity score indicates greater uncertainty and poorer prediction. Thus, the quality and relevance of the training corpus directly affect the performance of the language model.
