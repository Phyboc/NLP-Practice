```python
import nltk
from nltk.parse import ChartParser
from nltk import CFG

# -------------------------------------------------------------------
# Context-Free Grammar (CFG) Parsing & Ambiguity Detection
# -------------------------------------------------------------------
# Define a CFG with classic structural ambiguity (PP-attachment)
grammar = CFG.fromstring("""
  S -> NP VP
  PP -> P NP
  VP -> V NP | VP PP
  NP -> 'I' | Det N | NP PP
  Det -> 'the' | 'a'
  N -> 'man' | 'telescope' | 'dog'
  V -> 'saw'
  P -> 'with'
""")

parser = ChartParser(grammar)
sentence = "I saw the man with the telescope".split()

# Parse sentence to extract all valid parse trees
trees = list(parser.parse(sentence))

print(f"Sentence: {' '.join(sentence)}")
print(f"Total Parse Trees Found: {len(trees)}")

# A sentence is ambiguous under a CFG if strictly more than one parse tree exists
is_ambiguous = len(trees) > 1
print(f"Is Ambiguous? {is_ambiguous}\n")

# Display each parse tree ASCII visualization
for i, tree in enumerate(trees, 1):
    print(f"--- Parse Tree {i} ---")
    tree.pretty_print()
```
