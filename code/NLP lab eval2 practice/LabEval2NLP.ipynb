````python
import sqlite3
import pandas as pd

# Create a small SQLite database in the same folder as this notebook
conn = sqlite3.connect('auth_group.db')
cur = conn.cursor()
cur.execute('DROP TABLE IF EXISTS auth_group')
cur.execute('CREATE TABLE auth_group (id INTEGER PRIMARY KEY, name TEXT NOT NULL UNIQUE)')
rows = [
    (1, 'bike_admin'),
    (2, 'sales_manager'),
    (3, 'inventory_staff'),
    (4, 'finance_officer'),
    (5, 'customer_support'),
    (6, 'service_manager'),
]
cur.executemany('INSERT INTO auth_group (id, name) VALUES (?, ?)', rows)
conn.commit()
print(pd.read_sql_query('SELECT * FROM auth_group ORDER BY id', conn))
conn.close()
````

### Generate a sample bike-showroom NER corpus
The data below is generated for a bike sales showroom and contains examples of entities such as names, bike models, locations, and organizations.


````python
sample_texts = [
    'Suresh sold a Royal Enfield Classic 350 to Rahul in Bengaluru.',
    'The bike_admin approved the finance for a Bajaj Pulsar 150 at the showroom in Mysuru.',
    'TVS Motor Company provided servicing for the Apache RTR 160 in Coimbatore.',
    'Riya contacted customer_support to book a Honda Unicorn 160 in Kochi.',
    'The sales_manager closed a deal for KTM Duke 390 with Amit in Hyderabad.'
]

for text in sample_texts:
    print(text)
    print('-' * 80)
````

````python
from transformers import pipeline

# BERT NER model
bert_ner = pipeline(
    'token-classification',
    model='dbmdz/bert-large-cased-finetuned-conll03-english',
    aggregation_strategy='simple'
)

# DistilBERT NER model
distilbert_ner = pipeline(
    'token-classification',
    model='dbmdz/distilbert-large-cased-finetuned-conll03-english',
    aggregation_strategy='simple'
)

sample_texts = [
    'Suresh sold a Royal Enfield Classic 350 to Rahul in Bengaluru.',
    'The bike_admin approved the finance for a Bajaj Pulsar 150 at the showroom in Mysuru.',
    'TVS Motor Company provided servicing for the Apache RTR 160 in Coimbatore.',
    'Riya contacted customer_support to book a Honda Unicorn 160 in Kochi.'
]

for text in sample_texts:
    print('\nTEXT:', text)
    print('BERT ->', bert_ner(text))
    print('DistilBERT ->', distilbert_ner(text))
````
